```python
import pandas as pd
import matplotlib.pyplot as plt

# Leer CSV
df = pd.read_csv('./resultados/resultado_memoria.csv')

# Gráfico simple
plt.figure(figsize=(10, 6))
plt.plot(df['i'], df['nodes']/df['characters'], 'o-', linewidth=2, markersize=8)
plt.xlabel(r'n (Donde $i=2^n$)')
plt.ylabel('Número de Nodos Normalizado (nodos/caracteres)')
plt.title(r'Cantidad de Nodos del Trie en función de $2^n$ palabras')
plt.grid(True, alpha=0.3)
plt.show()
```
