```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('./resultados/resultado_memoria.csv')

# Gráfico de memoria
plt.figure(figsize=(8, 5))
plt.plot(df['i'], df['nodes']/df['characters'], 'o-', linewidth=2, markersize=8)
plt.xlabel(r'n (Donde $i=2^n$)')
plt.ylabel('Número de Nodos Normalizado (nodos/caracteres)')
plt.title(r'Cantidad de Nodos del Trie en función de $2^n$ palabras')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
df = pd.read_csv('./resultados/resultado_tiempo.csv')

# Gráfico de tiempo
plt.figure(figsize=(8, 5))
plt.plot(df['N/M'], df['t']/df['characters'], 'o-', linewidth=2, markersize=8)
plt.xlabel(r'#$\frac{N}{M}$')
plt.ylabel('Tiempo de insersión por carácter (ns)')
plt.title(r'Evolución temporal de inserción por carácter')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
df = pd.read_csv('./resultados/resultado_autocompletado.csv')

df_frecuencia = df[df['variante']=='Reciente']
# Gráfico de autocompletadp
plt.figure(figsize=(8, 5))
plt.plot(df_frecuencia['i'], df_frecuencia['porcentaje'], 'o-', linewidth=2, markersize=8)
plt.xlabel(r'L')
plt.ylabel('Porcentaje')
plt.title(r'Porcentaje de caracteres escritos c/r a los totales')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
df = pd.read_csv('./resultados/resultado_autocompletado.csv')

df_wiki_f = df[(df['dataset']=='wikipedia') & (df['variante']=='Frecuencia')]
df_wiki_r = df[(df['dataset']=='wikipedia') & (df['variante']=='Reciente')]


fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(10, 8))
axes[0].plot(df_wiki_f['i'], df_wiki_f['porcentaje'], 'o-', linewidth=2, markersize=8)
axes[0].set_title('Variante frecuencia')
axes[0].grid(True, alpha=0.3)

axes[1].plot(df_wiki_r['i'], df_wiki_r['porcentaje'], 'o-', linewidth=2, markersize=8)
axes[1].set_title('Variante reciente')
axes[1].grid(True, alpha=0.3)

fig.suptitle(r'Porcentaje de caracteres escritos c/r a los totales; Dataset = wikipedia.txt')
plt.xlabel(r'L')
plt.ylabel('Porcentaje')
plt.tight_layout() # Ajusta el espaciado
plt.show()
```

```python
df_random_f = df[(df['dataset']=='random') & (df['variante']=='Frecuencia')]
df_random_r = df[(df['dataset']=='random') & (df['variante']=='Reciente')]


fig, axes = plt.subplots(nrows=2, ncols=1, figsize=(10, 8))
axes[0].plot(df_random_f['i'], df_random_f['porcentaje'], 'o-', linewidth=2, markersize=8)
axes[0].set_title('Variante frecuencia')
axes[0].grid(True, alpha=0.3)

axes[1].plot(df_random_r['i'], df_random_r['porcentaje'], 'o-', linewidth=2, markersize=8)
axes[1].set_title('Variante reciente')
axes[1].grid(True, alpha=0.3)

fig.suptitle(r'Porcentaje de caracteres escritos c/r a los totales; Dataset = random.txt')
plt.xlabel(r'L')
plt.ylabel('Porcentaje')
plt.tight_layout() # Ajusta el espaciado
plt.show()
```
